# N2 physisorption: Micromeritics TriStar II 3020

`from_tristar_xls` reads the MicroActive Excel export into two
`PhysisorptionData` objects, one per branch of the isotherm. The BET result is a
typed attribute; everything else the instrument reports is kept under `report`.
The files are the test fixtures under `tests/physisorption/tristar/`.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

import lab_data_parser as ldp
```

```python
XLS = sorted((Path("..") / "tests" / "physisorption" / "tristar").glob("*.XLS"))[0]
branches = ldp.PhysisorptionData.from_tristar_xls(XLS)
adsorption, desorption = branches["adsorption"], branches["desorption"]
print(adsorption.n_points, "adsorption and", desorption.n_points, "desorption points")
```

```python
fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(adsorption.relative_pressure, adsorption.values, "o-", ms=3, label="adsorption")
ax.plot(desorption.relative_pressure, desorption.values, "s-", ms=3, label="desorption")
ax.set_xlabel("relative pressure / –")
ax.set_ylabel("quantity adsorbed / cm^3 g^-1 (STP)")
ax.legend()
plt.show()
```
